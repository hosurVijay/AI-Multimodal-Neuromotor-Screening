```python
def check_dataset(files, feature_cols):
    result = []
    for file in files:
        df = pd.read_csv(file, low_memory=False)
        missing_col = [col for col in feature_cols if col not in df.columns]

        if missing_col:
            result.append({
                'file': file.name,
                'missing_col': missing_col,
                'has_nan':None,
                'has_inf': None
            })
            continue

        sensor_df = df[feature_cols].apply(pd.to_numeric, errors = "coerce")

        result.append({
            "file": file.name,
            "missing_col": [],
            "has_nan": sensor_df.isnull().any().any(),
            "has_inf" : np.isinf(sensor_df).any().any()
        })

    return pd.DataFrame(result)
```

```python
pdpath = Path("../dataset/balanceDataset/PD")
pdFiles = list(pdpath.glob("*.csv"))
```

```python
print("PD", len(pdFiles))
```

```python
healthyCheck = check_dataset(filesHealthy, balance_features)
pdCheck = check_dataset(pdFiles, balance_features)
```
